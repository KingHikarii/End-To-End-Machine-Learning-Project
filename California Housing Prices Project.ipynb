**Creating the Test Set**

We will be setting aside 20% of dataset in order to check how accurate our  machine model is.

```python
import numpy as np

def split_train_test(data, test_ratio):
  shuffled_indices = np.random.permutation(len(data))

  test_size = int(len(data) * test_ratio)
  test_indices = shuffled_indices[:test_size]
  train_indices = shuffled_indices[test_size:]
  return data.iloc[train_indices], data.iloc[test_indices]

train_set, test_set = split_train_test(housing_data, 0.2)

len(train_set), len(test_set)
```

We need to compute hash of each instance's identifier and put that instance in the test set IF THE HASH IS LOWER THAN 20% of the maximum hash value. This ensures that the test set will remain consistent across multiple runs, because the previous coding will end up breaking at one point because it is using a different test set!

**Computationation of Hashes Concerning Each Instances Identifiers**


```python
from zlib import crc32

def test_set_check(identifier, test_ratio):
  return crc32(np.int64(identifier)) & 0xffffffff < test_ratio * 2**32
```

```python
def split_train_test_by_id(data, test_ratio, id_column):
  ids = data[id_column]
  in_test_set = ids.apply(lambda id_: test_set_check(id_, test_ratio))
  return data.loc[~in_test_set], data.loc[in_test_set]
```

Since the housing data does not have an identifer column, we will use the row index as the idenfifiers.

```python
housing_with_id = housing_data.reset_index()
train_set, test_set = split_train_test_by_id(housing_with_id, 0.2, "index")
```

**Custom Transformers**

In Scikit-Learn, a *transformer* is any object with `fit()` and `transform()` methods that changes the data, such as `SimpleImputer` or `StandardScaler`. (This is unrelated to the Transformer neural network architecture used in language models.) Scikit-Learn provides many transformers, but sometimes we need our own for custom cleanup operations or for combining specific attributes. We will write a small transformer class that adds the combined attributes we created earlier.

Inheriting from `TransformerMixin` gives us `fit_transform()` for free, and inheriting from `BaseEstimator` gives us `get_params()` and `set_params()`, which are needed for hyperparameter tuning.

```python
from sklearn.base import BaseEstimator, TransformerMixin

rooms_ix, bedrooms_ix, population_ix, households_ix = 3, 4, 5, 6

class CombinedAttributesAdder(BaseEstimator, TransformerMixin):
  def __init__(self, add_bedrooms_per_room = True):
    self.add_bedrooms_per_room = add_bedrooms_per_room
  def fit(self, X, y = None):
    return self
  def transform(self, X):
    rooms_per_household = X[:, rooms_ix] / X[:, households_ix]
    population_per_household = X[:, population_ix] / X[:, households_ix]
    if self.add_bedrooms_per_room:
      bedrooms_per_room = X[:, bedrooms_ix] / X[:, rooms_ix]
      return np.c_[X, rooms_per_household, population_per_household, bedrooms_per_room]
    else:
      return np.c_[X, rooms_per_household, population_per_household]

attr_adder = CombinedAttributesAdder(add_bedrooms_per_room=False)
housing_extra_attribs = attr_adder.transform(housing_data.values)
```

Variables that can be configured are known as hyperparameters. Hyperparameters control the learning process, model structure, function, and performance. These hyperparameters are set by the user before the training has begun and are external to the model, causing it not be changed during training.

In our case, our transformer has only one hyperparameter, *add_bedrooms_per_room*, which defaults to *True* (above we set it to *False* to try it out), by adding this hyperparameter, it will help to know whether this attribute helps our algorithm or not.

**Feature Scaling** - Machine learning algorithms don't perform well when the input numerical attributes have different scales. There are two common ways to get all the attributes to have the same scale and that is by using min-max scaling and standardization.

**Standardization** - Different from min-max scaling, standardization does not bound values to a specific range and it much less effected by outliers. We will be writing down the code to do this in the form of **transformation pipelines**

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy = 'median')),
    ('attribs_adder', CombinedAttributesAdder()),
    ('std_scaler', StandardScaler()), ])

housing_num_tr = num_pipeline.fit_transform(housing_num)
```

### Numerical Preprocessing Pipeline

A `Pipeline` chains the preprocessing steps for the **numerical features** so they always run in the same order:

1. **`imputer`**: `SimpleImputer(strategy='median')` fills missing values with each column's median, which is robust to outliers.
2. **`attribs_adder`**: `CombinedAttributesAdder()` is a custom transformer that adds engineered features, such as rooms per household, population per household, and bedrooms per room.
3. **`std_scaler`**: `StandardScaler()` standardizes every feature to mean 0 and standard deviation 1.

`fit_transform(housing_num)` fits each step on the data and transforms it before passing the result to the next step. It returns a NumPy array.

**Notes:**
- Every step except the last must be a transformer, meaning it has `fit()` and `transform()`.
- The pipeline's methods mirror the final step's methods.
- Use `fit_transform()` on the training data and only `transform()` on the test or new data, to avoid data leakage.
- You can access a step by name, e.g. `num_pipeline.named_steps['imputer']`, or tune it with grid-search parameters like `imputer__strategy`.


```python
from sklearn.compose import ColumnTransformer

num_attribs = list(housing_num)
cat_attribs = ['ocean_proximity']

full_pipeline = ColumnTransformer([
    ('num', num_pipeline, num_attribs),
    ('cat', OneHotEncoder(), cat_attribs),
])

housing_prepared_data = full_pipeline.fit_transform(housing_data)
```

### Full Preprocessing with `ColumnTransformer`

`ColumnTransformer` applies **different transformations to different columns** and concatenates the results into a single output.

- `num_attribs = list(housing_num)`: the names of all numerical columns.
- `cat_attribs = ['ocean_proximity']`: the categorical column.

Each entry is a `(name, transformer, columns)` tuple:

| Name | Transformer | Columns | What it does |
|------|-------------|---------|--------------|
| `num` | `num_pipeline` | `num_attribs` | Impute → add attributes → scale |
| `cat` | `OneHotEncoder()` | `cat_attribs` | One binary column per category |

`fit_transform(housing_data)` fits and transforms each column group, then joins the results side by side, ready for model training.

**Notes:**
- Columns that aren't listed are **dropped** by default (`remainder='drop'`). Use `remainder='passthrough'` to keep them.
- The output can be **sparse or dense**, depending on how many of its values are zero (`sparse_threshold=0.3`).
- `OneHotEncoder` raises an error on unseen categories. Use `handle_unknown='ignore'` if new data might contain them.
- `housing_data` should contain the features only, without the label `median_house_value`.

**Linear Regression**

Linear Regression is a statistical model that predicts an output (ŷ) as a weighted sum of the input features plus a constant called the **bias term** (or **intercept**):

ŷ = θ₀ + θ₁x₁ + θ₂x₂ + ... + θₙxₙ

- θ₀ is the bias term: the prediction when all features are 0.
- θ₁ ... θₙ are the weights the model learns for each feature.

`lin_reg.fit(housing_prepared_data, housing_labels)` trains the model:
- `housing_prepared_data` (X) holds the prepared input features.
- `housing_labels` (y) holds the true median house values.

During training, the model finds the weights that minimize the error between its predictions and the true labels. By default it minimizes the Mean Squared Error, the average of the squared differences.

**Measuring the Model's Error with RMSE**

To evaluate the model on the **whole training set**, we compute the **Root Mean Squared Error (RMSE)**:

- `lin_reg.predict(housing_prepared_data)` predicts a house value for every training instance.
- `mean_squared_error(housing_labels, housing_predictions_lin)` averages the squared differences between the true and predicted values (MSE).
- `np.sqrt(lin_mse)` takes the square root, which puts the error back in the same units as the target (dollars).

RMSE = √( (1/m) Σ (ŷᵢ − yᵢ)² )

Squaring the errors penalizes large mistakes more heavily than small ones.

**Result:** RMSE ≈ **$68,628**

On average, the model's predictions are off by about $68,628. Most districts' median house values fall between roughly $120,000 and $265,000, so this error is large.

**Interpretation: the model is underfitting the training data.** Either the features don't contain enough information to make good predictions, or the model isn't powerful enough. The main ways to fix underfitting are:
1. Choose a more powerful model.
2. Give the algorithm better features.
3. Reduce constraints on the model. (This model isn't regularized, so this option doesn't apply here.)

**Decision Tree Regressor**

A Decision Tree is an algorithm that makes predictions by asking a series of True/False questions about the features (for example, *"Is median_income ≤ 5.0?"*). It is made up of different nodes:

- **Root Node**: the starting point of the tree and its first question, asked about the whole dataset.
- **Decision (Internal) Nodes**: nodes that ask a question and split the data into two groups.
- **Branches**: the paths leaving a node, one for True and one for False.
- **Leaf Nodes**: the end points that give the final prediction. In a regression tree, the prediction is the **average target value** of the training instances in that leaf.

**How the tree is built:** At each node, the algorithm tests many possible features and thresholds. It picks the split that makes the resulting groups most similar in house value, which for regression means minimizing the **MSE**. It repeats this until a stopping condition is reached.

**Strengths:**
- Can capture complex, non-linear relationships.
- Doesn't require feature scaling.
- Easy to visualize and interpret.

**Weakness:** Without limits such as `max_depth` or `min_samples_leaf`, a tree can grow until it **memorizes** the training data. This is called overfitting: the model performs perfectly on training data but poorly on new data.

According to our trees' root mean squared error, we have an error of 0, which indicates that the model has badly overfitted the data. When it comes to machine learning, you do not want to touch the test set until you are ready to launch a model you are confident about, so you should use part of the training set for training and part of it for model validation.

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(tree_reg, housing_prepared_data, housing_labels,
                         scoring = "neg_mean_squared_error", cv = 10)
tree_rmse_scores = np.sqrt(-scores)
```

```python
def display_scores(scores):
    print("Scores: ", scores)
    print("Mean: ", scores.mean())
    print("Standard Deviation: ", scores.std())

display_scores(tree_rmse_scores)
```

**Evaluating the Decision Tree with Cross-Validation**

The Decision Tree's RMSE on the training set was 0.0. That doesn't mean the model is perfect: it **memorized** the training data (overfitting). To see how it performs on data it hasn't seen, we use **K-fold cross-validation**.

**How it works (`cv=10`):**
1. The training set is split into 10 equal parts called **folds**.
2. The model is trained 10 times. Each time, it trains on 9 folds and is evaluated on the 1 fold it didn't see.
3. The result is 10 scores, one for each fold.

**Why `scoring="neg_mean_squared_error"`?**
Scikit-Learn's cross-validation expects a **utility function** (greater is better) rather than a cost function (lower is better). So it returns the *negative* MSE, and we use `np.sqrt(-scores)` to flip the sign back and convert MSE into RMSE (in dollars).

**`display_scores()`** prints the 10 RMSE scores, their mean, and their standard deviation.

**Results:**
- **Mean RMSE ≈ $71,269**: the typical prediction error on unseen data.
- **Standard deviation ≈ $2,692**: how much the error varies across folds. This tells us how **precise** our estimate of the error is.

**Interpretation:**
The Decision Tree's perfect score on the training set was misleading. On unseen data, it has an error of about **$71,269 ± $2,692**, which is no better than (and possibly worse than) the Linear Regression model. The tree is **overfitting** so badly that it performs worse than a simpler model.

Cross-validation gives a more reliable estimate of model performance and also tells us how precise that estimate is. The cost is that the model has to be trained several times, which isn't always possible.

**Cross-Validating the Linear Regression Model**

Now that `cross_val_score` and `display_scores()` are defined, we run the same 10-fold cross-validation on the Linear Regression model so we can compare it with the Decision Tree.

```python
lin_scores = cross_val_score(lin_reg, housing_prepared_data, housing_labels,
                             scoring="neg_mean_squared_error", cv=10)
lin_rmse_scores = np.sqrt(-lin_scores)
display_scores(lin_rmse_scores)
```

**Random Forest Regressor**

A **Random Forest** trains many Decision Trees (100 by default), each on a random subset of the training data and using random subsets of features when splitting. Its prediction is the **average** of all the trees' predictions.

Building a model on top of many other models is called **Ensemble Learning**. Individual trees overfit in different ways, and averaging them cancels out much of that error. The result is a model that generalizes better than any single tree.

**Results:**
- **Training RMSE ≈ $18,651**
- **Cross-validation RMSE ≈ $50,436 ± $2,203**

**Model comparison (cross-validation RMSE):**

| Model | Mean RMSE | Std. Dev. |
|-------|----------:|----------:|
| Linear Regression | ~$69,000 | $2,700 |
| Decision Tree | $71,269 | $2,692 |
| **Random Forest** | **$50,436** | **$2,203** |

The Random Forest is the most promising model so far: its error on unseen data is about **$20,000 lower** than the other two models, and its scores are the most consistent across folds.

**But it is still overfitting.** The training RMSE ($18,651) is much lower than the cross-validation RMSE (50,436). The model fits the training data far better than it predicts new data. Possible fixes:
1. **Simplify or constrain the model** with hyperparameters such as `max_depth`, `min_samples_leaf`, or `max_features`.
2. **Get more training data.**

**Next step:** Before spending too much time on one model, we should shortlist 2–5 promising models and then **fine-tune** them. We'll tune the Random Forest's hyperparameters with `GridSearchCV`.

## Saving the Model with joblib

Training a model takes time. The Random Forest alone has to build 100 decision trees. Once the notebook is closed, everything in memory is gone and the model would have to be trained again. Saving the trained model to a file means we can reload it later, or in another program, and make predictions right away without retraining.

**joblib.dump()** writes the model to a **.pkl** file and **joblib.load()** restores it exactly as it was. Scikit-Learn recommends joblib over Python’s built-in pickle because it stores the large NumPy arrays inside models more efficiently.

We save the preprocessing pipeline and the model together in a single Pipeline. The model was trained on prepared data: imputed, with extra attributes added, scaled and one-hot encoded. New data must go through exactly the same steps, using the values learned from the training set. Bundling the two means the saved file can take raw data, with **ocean_proximity** as text and missing values included, and return predictions directly. That rules out mistakes like forgetting a step or refitting the scaler on new data.

After loading, we predict the same 5 districts as before to confirm the saved model gives the same results as the one in memory.

# **Confidence interval: “How sure are we about the error?”**

When you test your final model on the test set, you get one number, say RMSE = 47,000. But the test set is just one sample of districts. With a slightly different set of districts, you might get 45,000 or $49,000.

A 95% confidence interval gives you a range instead of a single number, for example 45,000 to 49,000. It means: “we’re 95% sure the model’s real error is somewhere in this range.”

```python
from scipy import stats

confidence = 0.95
squared_errors = (final_predictions - y_test) ** 2
np.sqrt(stats.t.interval(confidence, len(squared_errors) - 1, loc = squared_errors.mean(), scale = stats.sem(squared_errors)))
```
